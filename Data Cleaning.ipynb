```python

# remove the letters before club name
data['Club']=data['Club'].str.strip()
data.head()
```

```python
# Clean Value, Wage, and Release Clause columns:
# convert currency strings like '€103.5M' / '€560K' into numeric floats
# by replacing M -> e6 and K -> e3 , then casting to float

for col in ['Value', 'Wage', 'Release Clause']:
    data[col] = (
        data[col]
        .str.replace('€', '', regex=False)
        .str.replace('M', 'e6', regex=False)
        .str.replace('K', 'e3', regex=False)
        .astype(float)
    )
```

```python
# Standardize Height and Weight into consistent numeric units (cm and kg),
# since the raw data mixes metric ('170cm') and imperial ('6'2"', '183lbs') formats

import re

def height_to_cm(val):
    if pd.isna(val):
        return None
    val = val.strip()
    if 'cm' in val:
        return float(val.replace('cm', ''))
    else:  # imperial format like 6'2"
        feet, inches = re.match(r"(\d+)'(\d+)\"", val).groups()
        return round(int(feet) * 30.48 + int(inches) * 2.54, 1)

def weight_to_kg(val):
    if pd.isna(val):
        return None
    val = val.strip()
    if 'kg' in val:
        return float(val.replace('kg', ''))
    else:  # imperial format like 183lbs
        lbs = float(val.replace('lbs', ''))
        return round(lbs * 0.453592, 1)

data['Height_cm'] = data['Height'].apply(height_to_cm)
data['Weight_kg'] = data['Weight'].apply(weight_to_kg)
```

```python
# Split Contract into structured columns: it has 3 formats —
# "2004 ~ 2021" (normal), "Free" (no club), "Jun 30, 2021 On Loan" (loan end date)

normal = data['Contract'].str.extract(r'^(\d{4}) ~ (\d{4})$')
data['Contract_Start'] = normal[0]
data['Contract_End'] = normal[1]

data['Contract_Status'] = 'Normal'
data.loc[data['Contract'] == 'Free', 'Contract_Status'] = 'Free'
data.loc[data['Contract'].str.contains('On Loan', na=False), 'Contract_Status'] = 'On Loan'
```

```python
# Convert star-rating columns ('4 ★', '5★', etc.) into plain integers 1-5
for col in ['W/F', 'SM', 'IR']:
    data[col] = data[col].str.extract(r'(\d)').astype(int)
```

```python
# Convert Hits to numeric (some values like '1.6K' use shorthand for thousands),
# then fill missing values with 0 (assuming no recorded hits = 0 hits)

data['Hits'] = data['Hits'].str.replace('K', 'e3', regex=False).astype(float)
data['Hits'] = data['Hits'].fillna(0)
```

```python
# Convert Loan Date End to a real date type.
# Null here doesn't mean "missing/unknown" - it means the player isn't on loan.
# Add an explicit On_Loan boolean so that meaning is obvious rather than ambiguous.

data['Loan Date End'] = pd.to_datetime(data['Loan Date End'], format='%b %d, %Y', errors='coerce')
data['On_Loan'] = data['Loan Date End'].notna()
```

```python
# Drop URL columns - not useful for stats analysis,name
data = data.drop(columns=['photoUrl','Name','playerUrl'])
```

```python
# Convert Joined from a text date like 'Jul 1, 2004' into an actual datetime type
data['Joined'] = pd.to_datetime(data['Joined'], format='%b %d, %Y', errors='coerce')
```
